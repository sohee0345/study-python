문자열을 다룰 때 자주 사용하는 메서드를 익히고, 문자열과 리스트가 어떻게 연결되는지 연습한다.

## 문제 1 — count()

다음 문자열에 `"a"`가 몇 개 들어 있는지 구하세요.

```python
text = "banana apple"
```

### 힌트
문자열 메서드 하나만 사용하면 됩니다.

상태: [ ]


````python
text = "banana apple"

# 여기에 코드를 작성하세요.
text.count("a")
````

## 문제 2 — split()

다음 문자열을 **공백 기준으로 나누어 리스트**로 만드세요.

```python
text = "python pandas numpy"
```

예상 결과:

```python
["python", "pandas", "numpy"]
```

### 힌트
`split()`

상태: [ ]


````python
text = "python pandas numpy"

# 여기에 코드를 작성하세요.
lst = text.split(" ")

print(lst)
````

## 문제 3 — replace()

다음 문자열의 `"python"`을 `"Python"`으로 변경하세요.

```python
text = "i like python"
```

예상 결과:

```text
i like Python
```

### 힌트
`replace(바꿀값, 새로운값)`

상태: [ ]


````python
text = "i like python"

# 여기에 코드를 작성하세요.
change = text.replace("python", "Python")

print(change)
````

## 문제 4 — strip()

다음 문자열의 **앞뒤 공백만 제거**하세요.

```python
text = "   hello python   "
```

예상 결과:

```text
hello python
```

### 힌트
문자열 안쪽의 공백은 그대로 남아야 합니다.

상태: [ ]


````python
text = "   hello python   "

# 여기에 코드를 작성하세요.
txt = text.strip()

print(txt)
````

## 문제 5 — join()

다음 리스트의 문자열을 하나의 문장으로 합치세요.

```python
words = ["Python", "is", "fun"]
```

예상 결과:

```text
Python is fun
```

### 힌트
`" ".join(...)`

상태: [ ]


````python
words = ["Python", "is", "fun"]

# 여기에 코드를 작성하세요.
word = " ".join(words)

print(word)
````

## 복습 문제 1 — 조건에 맞는 값의 개수 구하기

다음 리스트에서 **10보다 크면서 홀수인 숫자의 개수**를 구하세요.

```python
numbers = [4, 11, 15, 18, 21, 7, 30, 33]
```

예상 결과:

```text
4
```

### 힌트
조건을 만족할 때마다 개수를 1씩 증가시킨다.

상태: [ ]


````python
numbers = [4, 11, 15, 18, 21, 7, 30, 33]

# 여기에 코드를 작성하세요.
count = 0

for num in numbers:
    if num > 10:
        if num % 2 != 0:
            count += 1

print(count)
````

## 복습 문제 2 — 최솟값 직접 찾기

다음 리스트에서 **가장 작은 숫자**를 찾으세요.

```python
numbers = [14, 6, 19, 3, 11, 8]
```

이번에는 `min()`과 `sort()`를 사용하지 않습니다.

### 힌트
첫 번째 값을 기준값으로 정한 뒤, 더 작은 값이 나오면 기준값을 바꾼다.

상태: [ ]


````python
numbers = [14, 6, 19, 3, 11, 8]

# 여기에 코드를 작성하세요.
least = numbers[0]

for num in numbers:
    if num < least:
        least = num

print(least)
````

### 학습 메모
- 문자열 메서드 `count()`, `split()`, `replace()`, `strip()`, `join()` 사용법을 익혔다.
- 여러 조건을 검사할 때 중첩 `if`를 사용할 수 있고, `and`로 조건을 합칠 수도 있다.
- 최댓값/최솟값은 첫 번째 값을 기준값으로 두고 반복하면서 갱신할 수 있다.
