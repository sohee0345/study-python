## 문제 1

다음 리스트에서 **짝수만 출력**하세요.

```python
numbers = [3, 8, 11, 14, 20, 25, 30]
```

### 힌트
`for + if + %`

상태: [ ]


````python
numbers = [3, 8, 11, 14, 20, 25, 30]

# 여기에 코드를 작성하세요.
for i in numbers:
    if i % 2 == 0:
        print(i)
````

## 문제 2

다음 리스트에서 **10보다 큰 값만 `result` 리스트에 저장**하세요.

```python
numbers = [4, 12, 7, 19, 3, 15]
result = []
```

예상 결과:

```python
[12, 19, 15]
```

### 힌트
`append()`

상태: [ ]


````python
numbers = [4, 12, 7, 19, 3, 15]
result = []

# 여기에 코드를 작성하세요.
for i in numbers:
    if i > 10:
        result.append(i)

print(result)
````

## 문제 3

1부터 30까지 숫자 중 **3의 배수의 합**을 구하세요.

### 힌트

```python
total = 0
```

`range()`를 활용해보세요.

상태: [ ]


````python
total = 0

for i in range(1, 31):
    if i % 3 == 0:
        total += i

print(total)
# 여기에 코드를 작성하세요.
````

## 문제 4

다음 리스트에서 **홀수의 개수**를 구하세요.

```python
numbers = [1, 4, 7, 9, 10, 13, 18]
```

### 힌트
개수를 저장할 변수를 먼저 만드세요.

상태: [ ]


````python
numbers = [1, 4, 7, 9, 10, 13, 18]

# 여기에 코드를 작성하세요.
counts = 0

for i in numbers:
    if i % 2 != 0:
        counts += 1

print(counts)
````

## 문제 5

다음 리스트에서 **가장 큰 숫자**를 찾아 출력하세요.

```python
numbers = [5, 12, 3, 21, 8]
```

이번에는 `max()`를 사용하지 않습니다.

### 힌트
첫 번째 값을 기준값으로 잡고 하나씩 비교해보세요.

상태: [ ]


````python
numbers = [5, 12, 3, 21, 8]

numbers.sort()

print(numbers[-1])
````

````python
numbers = [5, 12, 3, 21, 8]

# 여기에 코드를 작성하세요.
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print(largest)
````

## 오답 메모
**4번**
- 홀수의 "값"을 저장하려고 했음
- 문제는 홀수의 "개수"를 구하는 것이었음
- 'count += 1' 방식으로 해결

**5번**
- 리스트를 `sort()`로 정렬한 뒤 마지막 값을 가져와 최댓값을 구했다.
- 결과는 동일하지만 정렬하지 않고 값을 하나씩 비교하면서 최댓값을 갱신하는 문제였다.
- 첫 번째 값을 기준값으로 잡고, 더 큰 값이 나오면 기준값을 바꾸는 방식으로 해결할 수 있다.
