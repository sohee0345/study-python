7. numbers = [1, 2, 3, 4, 5] 리스트가 있을 때, for 문을 이용해서 모든 숫자를 더한 합계를 구하고 출력하세요. (힌트: total = 0으로 시작해서 반복하면서 더해주세요.)

```python
numbers = [1, 2, 3, 4, 5]

total = 0

for num in numbers:
    total += num

print(total)
```

8. person = {"이름": "철수", "나이": 20, "직업": "학생"} 딕셔너리가 있을 때, for 문을 이용해서 모든 키와 값을 "이름: 철수" 형식으로 한 줄씩 출력하세요. (힌트: person.items()를 사용하면 키와 값을 동시에 꺼낼 수 있어요.)

```python
person = {"이름": "철수", "나이": 20, "직업": "학생"}

for name, age in person.items():
    print(f"{name}: {age}")
```

9. students = [{"이름": "철수", "점수": 90}, {"이름": "영희", "점수": 85}, {"이름": "민수", "점수": 70}] 리스트가 있을 때, for 문을 이용해서 각 학생의 이름과 점수를 "철수: 90점" 형식으로 한 줄씩 출력하세요.

```python
students = [{"이름": "철수", "점수": 90}, {"이름": "영희", "점수": 85}, {"이름": "민수", "점수": 70}]

for sd in students:
    print(f"{sd['이름']}: {sd['점수']}점")
```

10. 문제 9의 students 리스트에서, 점수가 80점 이상인 학생의 이름만 출력하세요. (힌트: for 문 안에서 if 조건문을 같이 사용하세요.)

```python
students = [{"이름": "철수", "점수": 90}, {"이름": "영희", "점수": 85}, {"이름": "민수", "점수": 70}]
for sd in students:
    if sd["점수"] >= 80:
        print(sd["이름"])
```
