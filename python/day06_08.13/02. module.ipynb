# URL 인코딩/디코딩

```python
from urllib import parse
```

```python
naver_str = "https://search.naver.com/search.naver?where=nexearch&sm=top_hty&fbm=0&ie=utf8&query=%EC%9D%B8%EA%B3%B5%EC%A7%80%EB%8A%A5&ackey=0ae6zde9"
```

```python
naver_str
```

```python
naver_url = parse.unquote(naver_str)
naver_url
```

# 정규표현식

```python
import re
```

```python
txt = "a1b2c3d4e5f6g8h8i9j0"

pat = re.compile("[a-zA-Z]") # 규칙: 영문만 추출하는 규칙
pat.findall(txt) # 찾아줘
```

```python
pat = re.compile("[0-9]") # 규칙: 숫자만 추출하는 규칙
pat.findall(txt)
```
