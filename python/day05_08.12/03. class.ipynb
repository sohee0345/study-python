# 클래스

```python
class 클래스명: # 틀(설계도) 정의
    def __init__(self, ...): # 객체 만들 때 채울 값 정의

객체 = 클래스명(값1, 값2, ...) # 틀로 객체 찍어내기

객체.속성 # 데이터 꺼내기
객체.메서드() # 행동 실행시키기

객체 = 클래스명() = # 이 틀로 새 인스턴스 하나 만들어서 이 변수에 담아라
```

````python
# 예시 1

class Student:

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_pass(self):
        return self.score >= 60

    def add_score(self, point):
        self.score += point
````

````python
s1 = Student("민수", 55)
s2 = Student("지영", 80)
````

````python
print(s1.name)
print(s1.score)
````

````python
print(s1.is_pass()) # False
s1.add_score(10)
print(s1.score)
print(s1.is_pass())
````

````python
students = [Student("민수", 65), Student("지영", 80), Student("철수", 40)]

for s in students:
    print(f"{s.name}: {'합격' if s.is_pass() else '불합격'}")
````
