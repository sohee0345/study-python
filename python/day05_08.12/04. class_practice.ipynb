문제 1. 기본 클래스 만들기

Person 클래스를 만드세요. name과 age를 받아서 저장하고, 자기소개() 메서드를 호출하면 "저는 OOO이고 OO살입니다"를 출력하게 하세요.

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def 자기소개(self):
        print(f"저는 {self.name}이고 {self.age}살입니다.")

s1 = Person("홍길동", 22)
s1.자기소개()
```

문제 2. 계산기 클래스

Calculator 클래스를 만드세요. num1, num2 두 숫자를 받아 저장하고, 더하기(), 빼기() 메서드가 각각 두 수의 합과 차를 반환하게 하세요.

```python
class Calculator:

    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def 더하기(self):
        return (self.num1 + self.num2)

    def 빼기(self):
        return (self.num1 - self.num2)

s1 = Calculator(3, 5)

print(s1.더하기(), s1.빼기())
```

문제 3. 통장 클래스

BankAccount 클래스를 만드세요. balance(잔액)를 받아 저장하고, 입금(금액) 메서드는 잔액을 늘리고, 출금(금액) 메서드는 잔액을 줄이되 잔액보다 많이 출금하려 하면 "잔액 부족"을 출력하게 하세요.

```python
class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def 입금(self, amount):
        self.balance = self.balance + amount

    def 출금(self, amount):
        if amount > self.balance:
            print("잔액 부족")

        else:
            self.balance = self.balance - amount

money = BankAccount(5000)

money.입금(1000)
print(money.balance)

money.출금(2000)
print(money.balance)

money.출금(100000)
```

문제 5. 조건 있는 메서드

Rectangle(직사각형) 클래스를 만드세요. width, height를 받아 저장하고, 넓이() 메서드는 넓이를 반환, is_square() 메서드는 정사각형이면 True를 반환하게 하세요.

```python
class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def extent(self):
        return (self.width * self.height)

    def is_square(self):
        if self.width == self.height:
            return True
        else:
            return False

r1 = Rectangle(5, 3)

print(r1.extent(), r1.is_square())
```

문제 6. 클래스 변수 활용

Car 클래스를 만드세요. count라는 클래스 변수(모든 객체가 공유)를 0으로 시작해서, 객체가 하나 생성될 때마다(__init__이 호출될 때마다) 1씩 늘어나게 하세요. brand는 인스턴스 변수로 따로 저장하세요. 3대 만들고 Car.count를 출력해서 3이 나오는지 확인해보세요.

문제 7. 여러 메서드가 서로 연결되는 경우

Wallet(지갑) 클래스를 만드세요. money를 받아 저장하고, 살수있나(가격) 메서드는 money가 가격보다 많거나 같으면 True, 아니면 False를 반환하게 하세요. 구매(가격) 메서드는 살수있나(가격)를 먼저 호출해서 True면 money를 그만큼 줄이고 "구매 완료" 출력, False면 "돈 부족" 출력하게 하세요. (힌트: 메서드 안에서 다른 메서드는 self.살수있나(가격)처럼 호출합니다.)

```python
class Wallet:

    def __init__(self, money):
        self.money = money

    def p_buy(self, price):
        if self.money >= price:
            return True
        else:
            return False

    def get(self, price):
        if self.p_buy(price):
            self.money = self.money - price
            print("구매 완료")
        else:
            print("잔액 부족")

w = Wallet(5000)

w.get(3000)
print(w.money)

w.get(5000)
print(w.money)
```

문제 8. 기본값 있는 매개변수

Player(게임 캐릭터) 클래스를 만드세요. name은 필수로 받고, hp(체력)는 기본값 100으로 설정하세요 (def __init__(self, name, hp=100):). 공격받기(데미지) 메서드는 hp를 데미지만큼 줄이고, hp가 0 이하가 되면 "사망"을, 아니면 남은 체력을 출력하게 하세요.

```python
class Player:

    def __init__(self, name, hp=100):
        self.name = name
        self.hp = hp

    def attacked(self, damage):
        self.hp = self.hp - damage
        if self.hp <= 0:
            print("사망")
        else:
            print(self.hp)

p = Player("토쿠노")

p.attacked(60)
```
