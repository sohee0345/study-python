collections

```python
from collections import Counter

lst = [
    "aa", "bb", "aa", "bb", "aa", "bb", "tt", "tt"
]

# 인스턴스화
# counter : 객체
counter = Counter(lst)
counter # Counter({'aa': 3, 'bb': 3, 'tt': 2})
type(counter) # collections.Counter
dict(counter) # {'aa': 3, 'bb': 3, 'tt': 2}
dict(counter)['aa'] # 3

# 객체 => 변수 & 함수
counter.elements() # <itertools.chain at 0x22e15959000>
list(counter.elements()) # ['aa', 'aa', 'aa', 'bb', 'bb', 'bb', 'tt', 'tt']
```

datetime

```python
from datetime import datetime

# now : 객체
now = datetime.now()
now # now.year, now.month, now.day, now.hour, now.minute, now.second, now.microsecond

now.year, now.month, now.day, now.hour, now.minute, now.second, now.microsecond

now.date()
now.time()
now.weekday()
now.strftime("%A %d. %B %Y") # 미국식 시간표현법
now.strftime("%Y-%m-%d %H:%M") # '2026-08-13 11:16'

str_datetime = '2026-08-13 11:16'
str_datetime

datetime.strptime(str_datetime, "%Y-%m-%d %H:%M")
datetime.strptime(str_datetime, "%Y-%m-%d %H:%M").year

from datetime import datetime, timedelta

dt1 = datetime(2016, 2, 19, 14)
dt2 = datetime(2016, 1, 2, 13)

td = dt1 - dt2

# 현재 시간
now = datetime.now()

diff = timedelta(days=1, hours=3, minutes=30) # s를 꼭 붙여야함
now - diff

from dateutil.parser import parse
parse('2026-08-13 11:13')
```
