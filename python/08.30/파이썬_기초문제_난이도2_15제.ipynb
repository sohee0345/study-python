## 1. 리스트에서 짝수만 골라내기

다음 리스트에서 **짝수만 골라 새로운 리스트**를 만들어 출력하세요.

```python
numbers = [3, 8, 11, 14, 20, 25, 30]
```

출력 예시:

```text
[8, 14, 20, 30]
```

**힌트:** `% 2`

---


````python
numbers = [3, 8, 11, 14, 20, 25, 30]

result = []

for num in numbers:
    if num % 2 == 0:
        result.append(num)

print(result)
````

## 2. 문자열에서 특정 문자 개수 세기

다음 문자열에서 문자 `"a"`가 몇 번 등장하는지 반복문을 사용해 구하세요.

```python
text = "banana apple"
```

출력 예시:

```text
4
```

**힌트:** `for`

---


````python
text = "banana apple"

count = 0

for cnt in text:
    if cnt == "a":
        count += 1

print(count)
````

## 3. 점수에 따라 등급 출력하기

변수 `score`가 다음과 같을 때 조건에 맞는 등급을 출력하세요.

- 90점 이상: A
- 80점 이상 90점 미만: B
- 70점 이상 80점 미만: C
- 70점 미만: D

```python
score = 83
```

출력:

```text
B
```

**힌트:** `if / elif / else`

---


````python
score = 83

if score >= 90:
    print("A")
elif 80 <= score < 90:
    print("B")
elif 70 <= score < 80:
    print("C")
else:
    print("D")
````

## 6. 딕셔너리에서 평균 점수 구하기

다음 딕셔너리에 학생별 점수가 저장되어 있습니다. 전체 학생의 평균 점수를 구하세요.

```python
scores = {
    "민수": 80,
    "지수": 95,
    "철수": 70,
    "영희": 85
}
```

출력:

```text
82.5
```

**힌트:** `.values()`

---


````python
scores = {
    "민수": 80,
    "지수": 95,
    "철수": 70,
    "영희": 85
}

score = scores.values()

total = sum(score)
student_count = len(score)

average = total / student_count

print(average)
````

## 7. 3의 배수 합 구하기

1부터 30까지 숫자 중 **3의 배수만 더한 결과**를 출력하세요.

단, 직접 숫자를 하나씩 더하지 말고 반복문을 사용하세요.

**힌트:** `range()`

---


````python
total = 0

for num in range(1, 31):
    if num % 3 == 0:
        total += num

print(total)
````

## 8. 입력값을 숫자로 안전하게 변환하기

정수로 변환할 수 없는 값이 입력되면:


```python

```
