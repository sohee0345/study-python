1. 정수 하나를 매개변수로 받아 그 수의 제곱을 반환하는 함수 square(n)을 작성하세요.

```python
def square(n):
    return n*n

print(square(4))
```

2. for문을 사용해 1부터 10까지의 숫자를 한 줄씩 출력하세요.

```python
for i in range(1, 11):
    print(i)
```

3. 리스트를 매개변수로 받아 모든 원소의 합을 반환하는 함수 sum_list(numbers)를 작성하세요. (for문 사용)

```python
def sum_list(numbers):
    total = 0
    for num in numbers:
        total = total + num
    return total

print(sum_list([4, 6, 7, 8, 9]))
```

4. for문을 사용해 1부터 20까지의 숫자 중 짝수만 출력하세요.

```python
for num in range(1, 21):
    if num % 2 == 0:
        print(num)
```

5. 이름을 매개변수로 받아 "안녕, {이름}!"을 반환하는 함수 greet(name)을 작성하고, 리스트 ["철수", "영희", "민수"]의 각 이름에 대해 for문으로 함수를 호출해 인사말을 출력하세요.

```python
def greet(name):
    return(f"안녕, {name}")

for name in ["철수", "영희", "민수"]:
    print(greet(name))
```
