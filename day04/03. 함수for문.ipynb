# 함수 + for문 문제

1. 리스트의 합을 반환하는 함수
- 숫자 리스트를 매개변수로 받아서 for문으로 모든 숫자를 더한 후, 그 결과를 return하는 list_sum(numbers) 함수를 만드세요.

```python
numbers = [1, 2, 3, 4, 5]
def list_sum(numbers):
    total = 0
    for i in numbers:
        total = total + i
    return total
list_sum(numbers)
```

2. 짝수의 합을 반환하는 함수
- 숫자 리스트를 받아서 짝수만 골라 모두 더한 후 반환하는 sum_even(numbers) 함수를 만드세요.

```python
numbers = [1, 2, 3, 4, 5, 6]
def sum_even(numbers):
    total = 0
    for i in numbers:
        if i % 2 == 0:
            total = total + i
    return total
sum_even(numbers)
```

3. 특정 숫자보다 큰 숫자의 개수
- 숫자 리스트와 기준 숫자를 매개변수로 받아서, 기준 숫자보다 큰 값이 몇 개인지 반환하는 count_greater(numbers, target) 함수를 만드세요.

```python
def count_greater(numbers, target):
    count = 0
    for number in numbers:
        if number > target:
            count = count + 1
    return count
count_greater([2, 7, 4, 9, 1, 8], 5)
```

4. 문자열에서 특정 글자의 개수
- 문자열과 찾고 싶은 글자를 매개변수로 받아서, 해당 글자가 몇 번 등장하는지 반환하는 count_char(text, char) 함수를 만드세요.

```python
def count_char(text, char):
    count = 0
    for letter in text:
        if letter == char:
            count = count + 1
    return count
count_char("banana", "a")
```

5. 리스트에서 가장 큰 숫자 찾기
- 숫자 리스트를 받아서 for문을 사용해 가장 큰 숫자를 반환하는 find_max(numbers) 함수를 만드세요.
- 단, max() 함수는 사용하지 마세요.

```python
def find_max(numbers):
    max_number = numbers[0]
    for number in numbers:
        if number > max_number:
            max_number = number

    return max_number
find_max([3, 8, 2, 10, 5])
```

6. 두 번째로 큰 숫자 찾기
- 숫자가 담긴 리스트를 받아서 두 번째로 큰 숫자를 반환하는 find_second_max(numbers) 함수를 만들어보세요.

```python
def find_second_max(numbers):
    max_num = numbers[0]
    sec_num = numbers[1]

    for num in numbers[2:]:
        if num > max_num:
            sec_num = max_num
            max_num = num
        elif num > sec_num:
            sec_num = num

    return sec_num

find_second_max([50, 5, 8, 20, 3])
```

7. 조건에 맞는 숫자의 평균
- 숫자 리스트를 받아서 짝수만 골라 평균을 반환하는 average_even(numbers) 함수를 만들어보세요.

```python
def average_even(numbers):
    total = 0
    count = 0
    for i in numbers:
        if i % 2 == 0:
            total += i
            count += 1
    return total / count

average_even([1, 2, 3, 4, 5, 6, 8])
```
