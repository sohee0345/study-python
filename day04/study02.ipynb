# for문

```python
numbers = [1, 2, 3, 4]

for number in numbers:
    print(number)
```

- for [변수] in [반복할 대상]:
    반복할_작업

- range(시작, 끝)

```python
for i in range(1,5):
    print(i)
```

1. 숫자 출력
- for문과 range()를 사용해서 1부터 5까지 출력하세요.

```python
for i in range(1,6):
    print(i)
```

2. 리스트 출력
- 다음 리스트의 모든 과일을 for문을 사용해서 하나씩 출력하세요.
- fruits = ["사과", "바나나", "포도", "딸기"]

```python
for fruits in ["사과", "바나나", "포도", "딸기"]:
    print(fruits)
```

3. 숫자의 합
- 다음 리스트의 모든 숫자를 for문을 사용해서 더한 후 결과를 출력하세요.
- numbers = [3, 5, 7, 9]

```python
numbers = [3, 5, 7, 9]

total = 0
for i in numbers:
    total = total + i
print(total)
```

4. 짝수만 출력
- for문을 사용해서 1부터 10까지의 숫자 중 짝수만 출력하세요.

```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

5. 문자열의 글자 개수 세기
- 다음 문자열에 알파벳이 총 몇 글자 있는지 for문을 사용해서 세어보세요.
- word = "Python"

6. 숫자의 합 구하기
- 다음 리스트의 모든 숫자를 for문을 사용해서 더하세요.
- numbers = [2, 4, 6, 8, 10]

```python
numbers = [2, 4, 6, 8, 10]
total = 0
for i in numbers:
    total = total + i
print(total)
```

2. 숫자의 개수 세기
- 다음 리스트에 숫자가 총 몇 개 있는지 for문을 사용해서 세세요.
- numbers = [10, 20, 30, 40, 50, 60]

```python
numbers = [10, 20, 30, 40, 50, 60]
count = 0
for i in numbers:
    count = count + 1
print(count)
```

3. 짝수의 합 구하기
- 다음 리스트에서 짝수만 골라서 모두 더하세요.
- numbers = [1, 2, 3, 4, 5, 6, 7, 8]

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
total = 0
for i in numbers:
    if i % 2 == 0:
        total = total + i
print(total)
```

4. 5보다 큰 숫자의 개수
- 다음 리스트에서 5보다 큰 숫자가 몇 개인지 세어보세요.
- numbers = [2, 7, 4, 9, 1, 8, 3, 6]

```python
numbers = [2, 7, 4, 9, 1, 8, 3, 6]
count = 0
for i in numbers:
    if i > 5:
        count = count + 1
print(count)
```

5. 문자열에서 특정 글자 찾기
- 다음 문자열에서 알파벳 "a"가 몇 번 나오는지 for문을 사용해서 세어보세요.
- word = "banana"

```python
word = "banana"
count = 0
for i in word:
    if i == "a":
        count = count + 1
print(count)
```
