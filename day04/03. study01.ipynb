### 문제 2: 모든 점을 x축 기준으로 대칭 이동시킨 새로운 좌표 리스트 생성

```python
## 문제 
- 모든 점을 x축 기준으로 대칭 이동시킨 새로운 좌표 리스트 생성
  - x축 대칭: (x, y) -> (x, -y)
```

````python
points = [(1, 2), (3, 4), (-1, 5), (2, -3), (0, 0), (4, 1), (-2, -2)]

reflected = [(x, -y) for x, y in points]
print(reflected)
````
